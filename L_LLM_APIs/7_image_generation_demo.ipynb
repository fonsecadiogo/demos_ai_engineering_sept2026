# Demo: Image Generation

```python
from openai import OpenAI
from pydantic import BaseModel, Field
from typing import Literal
import os
from dotenv import load_dotenv, find_dotenv

_ = load_dotenv(find_dotenv())  # read local .env file
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

client = OpenAI()
```

```python
from IPython.display import display, Image
import base64

result = client.images.generate(
    model="gpt-image-2.5-flare",
    prompt="A city where all the buildings are made of stacked books, aerial view",
    size="1024x1024",
    quality="medium",
    n=1,
)

image_bytes = base64.b64decode(result.data[0].b64_json)

display(Image(data=image_bytes))
```

## Experimenting with Different Prompts

**The more clearly you describe your goal, the more control you have over the generated image.** Start with a simple prompt, then experiment by adding details and refining the results.

Consider specifying the following elements:

- **Subject:** What should appear in the image? (e.g., a mountain climber, a futuristic city).
- **Style:** How should it look? (e.g., photorealistic, watercolor, digital illustration, pixel art).
- **Composition:** How should elements be arranged? (e.g., close-up, wide shot, overhead view, subject placement).
- **Lighting and colors:** What mood or atmosphere do you want? (e.g., golden-hour lighting, soft shadows, muted colors).
- **Context and background:** Where does the scene take place? What should surround the subject?
- **Purpose and audience:** What will the image be used for? (e.g., a marketing campaign, a presentation, a video game).

**Tip:** You don't need to specify everything in every prompt. Experiment with different levels of detail, compare the results, and refine your prompt based on what works best.

## API parameters

The `images.generate()` method supports options that affect the output.

| Parameter | What it controls |
|---|---|
| `model` | Which image-generation model to use. |
| `prompt` | Description of the image. |
| `size` | Image dimensions, such as `1024x1024`, `1536x1024`, or `1024x1536`. |
| `quality` | Rendering quality. Supported values depend on the model; this model supports `low`, `medium`, `high`, `xhigh`, and `max`. |
| `n` | Number of images to generate in one request. |
| `background` | Background behavior; supported GPT Image models can support `opaque` or `transparent`. |

Higher quality and multiple images may increase cost and/or generation time. Check the current documentation for supported combinations and pricing.
